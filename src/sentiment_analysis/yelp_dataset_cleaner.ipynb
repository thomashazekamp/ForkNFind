# Change column names for better reproducibility

```python
import pandas as pd


df = pd.read_csv('dataset/yelp_review_polarity_dataset/yelp_polarity_train.csv')

df.head(10)
```

```python
# Change the column names
old_name1 = df.columns[0]
df.rename(columns={old_name1: 'sentiment'}, inplace=True)

old_name = df.columns[1]

df.rename(columns={old_name: 'text'}, inplace=True)

df.head(10)
```

```python
# Convert the sentiment column to 'positive' and 'negative'
df['sentiment'] = df['sentiment'].map({1: 'negative', 2: 'positive'})
```

```python
df.to_csv('dataset/yelp_review_polarity_dataset/updated_yelp_polarity_train.csv', index=False)
```
