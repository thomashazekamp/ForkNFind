# Using Asserts to test sentiment Analysis working functionality

```python
# pick random 30 pieces of data from the yelp dataset
import pandas as pd

def random_select():
    df = pd.read_csv('../dataset/yelp_data/updated_yelp_polarity_train.csv')
    df = df.head(3) # pick top 3 rows
    df.to_csv('../dataset/yelp_data/top_3_train.csv', index=False) # save random 30 data to new csv file
    return df
```

```python
df = random_select()
```

```python
def assert_row(given_text, expected_sentiment): # test the sentiment analysis model
    from runner_sent_analysis import predict_sentiment

    pred_sent = predict_sentiment(given_text)

    # Testing that the sentiment analysis model works by producing a '1' or '0'
    print("Running (functionality checks) asserts for sentiment analysis model...")
    assert pred_sent in ['1', '0']
    print("Test passed.")

    # Testing that the sentiment analysis model works by producing the expected sentiment
    print("Running (testing model - returning expected sentiment) asserts for sentiment analysis model...")
    if expected_sentiment == 'positive':
        assert pred_sent == '1'
    elif expected_sentiment == 'negative':
        assert pred_sent == '0'
    print("Test passed.")
```

```python

# Running the asserts on each row of the selected dataset
for row in df.itertuples():
    text = row.text
    sentiment = row.sentiment
    assert_row(text, sentiment)
```
